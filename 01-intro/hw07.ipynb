```python
list_dates = ['FECHA_MAT','FEC_INI_CURSO','FEC_FIN_CURSO','FEC_CERTIF']
```

```python
for col in list_dates:
    df[col] = pd.to_datetime(
        df[col],
        format='%Y%m%d',
        errors='coerce'
    )
```

```python
#FILL missing values in 'TIPO_CERTIF' with NA
df['TIPO_CERTIF'] = df['TIPO_CERTIF'].fillna('NA')
```

```python
df.head()
```

```python
df['pago'] = (df['OBTUVO_CERTIF'] == 'SI').astype(int)
```

```python
df['pago'].value_counts()
```

```python
#Creando la variable N° de edicion 
df['N_EDICION'] = np.nan
# Primera edición (2024)
df.loc[
    (df['FECHA_MAT']>= '2024-05-14') & (df['FEC_INI_CURSO']< '2024-07-08'),
    'N_EDICION'
] = 1

# Segunda edición (2024)
df.loc[
    (df['FECHA_MAT']>= '2024-07-08') & (df['FEC_INI_CURSO']< '2024-10-09'),
    'N_EDICION'
] = 2

# Tercera edición (2024)
df.loc[
    (df['FECHA_MAT']>= '2024-10-10') & (df['FEC_INI_CURSO']< '2025-01-08'),
    'N_EDICION'
] = 3

# Cuarta edición (2025)
df.loc[
    (df['FECHA_MAT']>= '2025-01-09') & (df['FEC_INI_CURSO']< '2025-04-20'),
    'N_EDICION'
] = 4

# Quinta edición (2025)
df.loc[
    (df['FECHA_MAT']>= '2025-04-21') & (df['FEC_INI_CURSO']<= '2025-06-24'),
    'N_EDICION'
] = 5
```

```python
df.sample(10)
```

```python
df['N_EDICION'] = df['N_EDICION'].astype('Int64')
```

```python
for i in range(1, 6):
    print(f"Edición {i}:")
    print(df[df.N_EDICION==i]['pago'].value_counts())
    print()
#df[df.N_EDICION==1]['pago'].value_counts(normalize=True)
```
